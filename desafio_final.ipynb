```python
def ler_transacoes(CAMINHO_ARQUIVO: str) -> list[dict]:
  transacoes = []
  try:
    with open (CAMINHO_ARQUIVO, mode = "r", encoding="utf-8") as f:
      leitor = csv.DictReader(f)
      for linha in leitor:
        transacoes.append(linha)
  except FileNotFoundError:
    print(f"Erro: o arquivo '{CAMINHO_ARQUIVO}' não fo encontrado;")
    return[]
  return transacoes
```

```python
from datetime import datetime

def validar_transacao(linha:dict) -> dict | None:
  try:
    id_str = linha.get("id", "").strip()
    if not id_str:
      return None
    id_valido=int(id_str)

    cliente_id = linha.get("cliente_id", "").strip()
    if not cliente_id:
      return None

    tipo = linha.get("tipo","").strip().lower()
    if tipo not in {"credito","debito"}:
      return None

    data_str = linha.get("data","").strip()
    data_objeto = datetime.strptime(data_str,"%Y-%m-%d")

    valor_str = linha.get("valor", "").strip()
    valor_float=float(valor_str)
    if valor_float <= 0:
      return None
    return {
        "id":id_valido,
        "data":data_objeto,
        "cliente_id":cliente_id,
        "tipo":tipo,
        "valor":valor_float,
        "descricao":linha.get("descricao","").strip(),
        "categoria":linha.get("categoria","").strip()
    }
  except (ValueError, TypeError, KeyError):
    return None

```

```python
from datetime import datetime

LIMITE_SUSPEITO = 10000.00


def gerar_relatorio(transacoes_validas: list[dict]) -> dict:

    if not transacoes_validas:
        return {
            "resumo_mensal": {},
            "suspeitas": [],
            "periodo": {"data_inicio": None, "data_fim": None, "dias_totais": 0},
        }


    dados_agrupados = {}
    transacoes_suspeitas = []

    todas_datas = []

    for tx in transacoes_validas:
        data_tx = tx["data"]
        valor_tx = tx["valor"]
        tipo_tx = tx["tipo"]

        todas_datas.append(data_tx)

        if valor_tx > LIMITE_SUSPEITO:
            transacoes_suspeitas.append({
                "id": tx["id"],
                "cliente_id": tx["cliente_id"],
                "data": data_tx.strftime("%Y-%m-%d"),
                "valor": valor_tx,
            })
        mes_ano = data_tx.strftime("%Y-%m")

        if mes_ano not in dados_agrupados:
            dados_agrupados[mes_ano] = {
                "quantidade": 0,
                "total_credito": 0.0,
                "total_debito": 0.0,
                "valores": [],
            }

        dados_agrupados[mes_ano]["quantidade"] += 1
        dados_agrupados[mes_ano]["valores"].append(valor_tx)

        if tipo_tx == "credito":
            dados_agrupados[mes_ano]["total_credito"] += valor_tx
        elif tipo_tx == "debito":
            dados_agrupados[mes_ano]["total_debito"] += valor_tx

    resumo_mensal = {}

    for mes in sorted(dados_agrupados.keys()):
        info = dados_agrupados[mes]
        valores = info["valores"]
        tot_credito = round(info["total_credito"], 2)
        tot_debito = round(info["total_debito"], 2)

        resumo_mensal[mes] = {
            "quantidade": info["quantidade"],
            "total_credito": tot_credito,
            "total_debito": tot_debito,
            "saldo": round(tot_credito - tot_debito, 2),
            "media": round(sum(valores) / len(valores), 2),
            "maior_valor": max(valores),
            "menor_valor": min(valores),
        }

    data_min = min(todas_datas)
    data_max = max(todas_datas)
    dias_totais = (data_max - data_min).days

    return {
        "resumo_mensal": resumo_mensal,
        "suspeitas": transacoes_suspeitas,
        "periodo": {
            "data_inicio": data_min.strftime("%Y-%m-%d"),
            "data_fim": data_max.strftime("%Y-%m-%d"),
            "dias_totais": dias_totais,
        },
    }
```

```python
def formatar_moeda(valor: float) -> str:

  return f"R${valor:,.2f}".replace(",","X").replace(".",",").replace("X",".")

def exibir_relatorio(relatorio: dict, total_lidas: int, total_validas: int, total_invalidas: int) -> None:

  print("\n" + "=" * 45)
  print("        CLEARBANK - RELATORIO FINANCEIRO")
  print("=" * 45)

  print(f"Total de linhas lidas:    {total_lidas}")
  print(f"Linhas válidas:           {total_validas}")
  print(f"Linhas inválidas:         {total_invalidas}")

  periodo = relatorio.get("periodo", {})
  data_inicio = periodo.get("data_inicio")
  data_fim = periodo.get("data_fim")
  dias = periodo.get("dias_totais", 0)

  if data_inicio and data_fim:
    print(f"Período analisado:         {data_inicio} - {data_fim} ({dias} dias)")
  else:
    print("Período analisado:        N/A")

  print("\n" + "=" * 45)
  print("               RESUMO MENSAL")
  print("=" * 45)

  resumo_mensal = relatorio.get("resumo_mensal", {})
  if not resumo_mensal:
    print("Nenhuma transação válida para ecibir no resumo mensal.")
  else:
    for mes, metricas in resumo_mensal.items():
      print(f"Mês: {mes}")
      print(f"  Transações:       {metricas['quantidade']}")
      print(f"  Total crédito:    {formatar_moeda(metricas['total_credito']):14}")
      print(f"  Total débito:     {formatar_moeda(metricas['total_debito']):14}")
      print(f"  Saldo:            {formatar_moeda(metricas['saldo']):14}")
      print(f"  Média:            {formatar_moeda(metricas['media']):14}")
      print(f"  Maior valor:      {formatar_moeda(metricas['maior_valor']):14}")
      print(f"  Menor valor:      {formatar_moeda(metricas['menor_valor']):14}")

  print("\n" + "=" * 45)
  print("           TRANSAÇÕES SUSPEITAS")
  print("=" * 45)

  suspeitas = relatorio.get("suspeitas", [])
  if suspeitas:
    for s in suspeitas:
      valor_fmt = formatar_moeda(s["valor"])
      print(f"ID: {s['id']:<3} | Cliente: {s['cliente_id']} | Valor: {valor_fmt}")
  else:
    print("Nenhuma transação suspeita encontrada.")
  print("=" * 45 + "\n")



```

```python
import json
from datetime import datetime

def salvar_json(relatorio: dict, total_validas: int, total_invalidas:int, caminho_saida: str = "relatorio.json") -> bool:
  dados_exportação = {
      "gerado_em:": datetime.now().strftime("%Y-%m-%d"),
      "total_transacoes_validas": total_validas,
      "total_transacoes_invalidas": total_invalidas,
      "periodo":relatorio.get("periodo",{}),
      "resumo_mensal":relatorio.get("resumo_mensal",{}),
      "transacoes_suspeitas":relatorio.get("suspeitas",[])
  }
  try:
    with open (caminho_saida, mode="w",encoding="utf-8") as f:
      json.dump(dados_exportação, f, ensure_ascii=False, indent=2)
    print(f"Arquivo exportado com sucesso: '{caminho_saida}'")
    return True
  except (IOError, OSError) as erro:
    print(f"Erro ao salvar arquivo JSON '{caminho_saida}': {erro}")
    return False
```

```python
def pipeline_analise_financeira(arquivo_origem: str = "transacoes.csv", arquivo_destino: str = "relatorio.json") -> None:
  print("Iniciando processamento do ClearBank...")

  transacoes_brutas = ler_transacoes(arquivo_origem)

  if not transacoes_brutas:
    print(f"Execução interrompida: nenhum registro encontrado em '{arquivo_origem}'")
    return

  transacoes_validas = []
  total_invalidas = 0

  for linha in transacoes_brutas:
    registro_limpo = validar_transacao(linha)
    if registro_limpo is not None:
      transacoes_validas.append(registro_limpo)
    else:
      total_invalidas += 1

  total_lidas = len(transacoes_brutas)
  total_validas = len(transacoes_validas)

  relatorio = gerar_relatorio(transacoes_validas)

  exibir_relatorio(
      relatorio=relatorio,
      total_lidas=total_lidas,
      total_validas=total_validas,
      total_invalidas=total_invalidas
  )

  salvar_json(
      relatorio=relatorio,
      total_validas=total_validas,
      total_invalidas=total_invalidas,
      caminho_saida=arquivo_destino
  )

pipeline_analise_financeira(CAMINHO_ARQUIVO, "relatorio.json")
```

```python
import pandas as pd
import numpy as np

def executar_analise_pandas(caminho_csv: str = "transacoes.csv"):

  print("\n" + "=" * 45)
  print("           CLEARBANK - ANALISE COM PANDAS")
  print("=" * 45)

  try:

    df = pd.read_csv(caminho_csv, dtype=str)
  except FileNotFoundError:
    print(f"Erro: Arquivo '{caminho_csv} não encontrado.")
    return

  linhas_totais = len(df)

  for col in df.columns:
    df[col] = df[col].astype(str).str.strip()

  df["id_num"] = pd.to_numeric(df["id"], errors="coerce")
  mascara_id = df["id_num"].notna() & (df["id"] != "")

  mascara_cliente = (df["cliente_id"] != "") & (
      df["cliente_id"].str.lower() != "nan"
  )

  df["tipo_norm"] = df["tipo"].str.lower()
  mascara_tipo = df["tipo_norm"].isin(["credito", "debito"])

  df["data_dt"] = pd.to_datetime(
      df["data"], format="%Y-%m-%d", errors="coerce")
  mascara_data = df["data_dt"].notna()

  df["valor_num"] = pd.to_numeric(df["valor"], errors="coerce")
  mascara_valor = df["valor_num"].notna() & (df["valor_num"] > 0)

  mascara_valida = (
      mascara_id & mascara_cliente & mascara_tipo & mascara_data & mascara_valor
  )
  df_limpo = df[mascara_valida].copy()

  linhas_validas = len(df_limpo)
  linhas_invalidas = linhas_totais - linhas_validas

  print(f"Total lido:    {linhas_totais}")
  print(f"Válidos:       {linhas_validas}")
  print(f"Descartados:   {linhas_invalidas}")

  df_limpo["mes"] = df_limpo["data_dt"].dt.strftime("%Y-%m")

  resumo = []
  for mes, grupo in df_limpo.groupby("mes"):
    qtd = len(grupo)
    tot_credito = grupo.loc[grupo["tipo_norm"] == "credito", "valor_num"].sum()
    tot_debito = grupo.loc[grupo["tipo_norm"] == "debito", "valor_num"].sum()
    saldo = tot_credito - tot_debito
    media = grupo["valor_num"].mean()
    maior = grupo["valor_num"].max()
    menor = grupo["valor_num"].min()

    resumo.append({
        "mes": mes,
        "total_credito": round(tot_credito, 2),
        "total_debito": round(tot_debito, 2),
        "saldo": round(saldo, 2),
        "media": round(media, 2),
        "maior_valor": round(maior, 2),
        "menor_valor": round(menor, 2),
    })

  df_resumo = pd.DataFrame(resumo)
  print("\n--- Resumo Mensal Agrupado (DataFrame) ---")
  print(df_resumo.to_string(index=False))

  suspeitas_pandas = df_limpo[df_limpo["valor_num"] > 10000.00]
  print(f"\nTransações suspeitas detectadas: {len(suspeitas_pandas)}")
  for _, linha in suspeitas_pandas.iterrows():
    print(f"ID: {linha['id']} | Cliente: {linha['cliente_id']} | Data: {linha['data']} | Valor: R$ {linha['valor_num']:,.2f}")
  return df_limpo, df_resumo

df_limpo, df_resumo = executar_analise_pandas("transacoes.csv")



```

```python
import matplotlib.pyplot as plt
import numpy as np

def gerar_grafico_financeiro(df_resumo: pd.DataFrame, caminho_imagem: str = "grafico.png"):
  meses = df_resumo["mes"].tolist()
  creditos = df_resumo["total_credito"].tolist()
  debitos = df_resumo["total_debito"].tolist()
  saldos = df_resumo["saldo"].tolist()
  x = np.arange(len(meses))
  largura = 0.25

  fig, ax = plt.subplots(figsize=(10,6), dpi=150)

  barras_credito = ax.bar(x - largura, creditos, largura, label="Total Crédito", color="#2ca02c")
  barras_debito = ax.bar(x, debitos, largura, label = "total Débitos", color="#d62728")
  barras_saldo = ax.bar(x + largura, saldos, largura, label="Saldo Liquido", color="#1f77b4")

  ax.axhline(0,color="gray", linewidth=0.8,linestyle="--")

  ax.set_title("Clearbank - Evolução financeira Mensal", fontsize=14, pad=15, fontweight="bold")
  ax.set_xlabel("Mês de referência", fontsize=11, labelpad=10)
  ax.set_ylabel("Valor (R$)", fontsize=11, labelpad=10)
  ax.set_xticks(x)
  ax.set_xticklabels(meses, fontsize=10)
  ax.legend(frameon=True, facecolor="white", edgecolor="none")
  ax.grid(axis="y",linestyle=":",alpha=0.6)

  def rotular_barras(barras):
    for barra in barras:
      altura = barra.get_height()
      deslocamento = 200 if altura >= 0 else -600
      ax.annotate(
          f"R$ {altura:,.0f}",
          xy=(barra.get_x() + barra.get_width() / 2, altura),
          xytext=(0, 3 if altura >= 0 else -10),
          textcoords="offset points",
          ha="center",
          va="bottom" if altura >= 0 else "top",
          fontsize=8,
          rotation=0
        )

  rotular_barras(barras_credito)
  rotular_barras(barras_debito)
  rotular_barras(barras_saldo)

  plt.tight_layout()

  plt.savefig(caminho_imagem, dpi=150)
  plt.show()
  print (f"Gráfico salvo com sucesso: '{caminho_imagem}'")

gerar_grafico_financeiro(df_resumo, "grafico.png")


```
